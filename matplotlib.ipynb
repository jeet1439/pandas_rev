```python
names = [
    "Rahul", "Priya", "Amit", "Sneha", "Arjun",
    "Neha", "Rohan", "Ananya", "Vikram", "Pooja",
    "Sourav", "Kavya", "Raj", "Meera", "Karan",
    "Isha", "Manish", "Tanya", "Deepak", "Riya"
]

ages = [
    25, 32, 45, 29, 51,
    38, 22, 41, 56, 34,
    47, 27, 63, 36, 43,
    31, 58, 24, 49, 39
]

blood_pressure = [
    118, 122, 135, 110, 145,
    128, 115, 130, 150, 120,
    138, 112, 155, 125, 132,
    117, 148, 108, 140, 126
]


colors = []

for bp in blood_pressure:
    if bp >= 140:
        colors.append("red")       # High
    elif bp >= 90:
        colors.append("green")     # Normal
    else:
        colors.append("blue")      # Low

# plt.scatter(ages, blood_pressure, c=colors)

plt.scatter(ages, blood_pressure, cmap="plasma", c=blood_pressure)
plt.xlabel("Age")

plt.ylabel("Systolic Blood Pressure")

plt.title("Age vs Blood Pressure")

plt.colorbar(label="blood pressure")
# plt.show()

for i in range(len(names)):
    plt.annotate(names[i], xy=(ages[i], blood_pressure[i]), xytext=(ages[i] + 2, blood_pressure[i]))

plt.xlim(min(ages) - 5, max(ages)+ 5)
```

```python
# City names
cities = [
    "Kolkata", "Delhi", "Mumbai", "Bangalore", "Chennai",
    "Hyderabad", "Pune", "Jaipur", "Lucknow", "Srinagar"
]

# Winter data
winter_temp = [15, 17, 12, 19, 14, 16, 11, 18, 13, 20]
winter_humidity = [65, 60, 72, 55, 68, 62, 75, 58, 70, 52]

# Summer data
summer_temp = [30, 33, 35, 32, 37, 34, 31, 38, 36, 39]
summer_humidity = [65, 60, 55, 62, 50, 58, 68, 48, 52, 45]

# Winter
plt.scatter(
    winter_temp,
    winter_humidity,
    color="blue",
    label="Winter"
)

# Summer
plt.scatter(
    summer_temp,
    summer_humidity,
    color="red",
    label="Summer"
)

# Add city names to each point
for i, city in enumerate(cities):
    plt.annotate(
        city,
        (winter_temp[i], winter_humidity[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

    plt.annotate(
        city,
        (summer_temp[i], summer_humidity[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.title("Temperature vs Humidity by City")

plt.legend()
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

direct_marketing = [120, 150, 130, 170, 180, 200, 190]
organic = [80, 90, 100, 110, 120, 130, 140]
social_media = [100, 120, 140, 130, 150, 160, 180]
email_marketing = [60, 70, 65, 80, 90, 100, 110]

plt.stackplot(
    days,
    direct_marketing,
    organic,
    social_media,
    email_marketing,
    labels=[
        "Direct Marketing",
        "Organic",
        "Social Media",
        "Email Marketing"
    ]
)

plt.xlabel("Days")
plt.ylabel("Visitors")
plt.title("Marketing Traffic")
plt.legend()

plt.show()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

sales = [120, 150, 130, 170, 180, 200, 190]
users = [80, 90, 100, 110, 120, 130, 140]
profit = [40, 50, 45, 60, 65, 70, 75]

fig, ax = plt.subplots(2, 2, figsize=(10, 7))

# 1. Line plot
ax[0, 0].plot(days, sales)
ax[0, 0].set_title("Sales")

# 2. Bar plot
ax[0, 1].bar(days, users)
ax[0, 1].set_title("Users")

# 3. Scatter plot
ax[1, 0].scatter(days, profit)
ax[1, 0].set_title("Profit")

# 4. Stack plot
ax[1, 1].stackplot(
    days,
    sales,
    users,
    profit,
    labels=["Sales", "Users", "Profit"]
)
ax[1, 1].set_title("Stack Plot")
ax[1, 1].legend()

plt.tight_layout()
plt.show()
```
