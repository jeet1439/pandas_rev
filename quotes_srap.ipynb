```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
for i in range(1, 10):
    res = requests.get(f"https://quotes.toscrape.com/tag/life/page/{i}/")
    
    if res.status_code == 200:
        print("Request successful")
    
        with open(f"quotes_scrap_data/data{i}.html", "w", encoding="utf-8") as f:
            f.write(res.text)
    else:
        print("Request failed:", res.status_code)  
```

```python
data_list = []

for i in range(1, 10):
    with open(f"quotes_scrap_data/data{i}.html", "r", encoding="utf-8") as f:
        raw_html = f.read()
        
    soup = BeautifulSoup(raw_html,"lxml")
    quotes = soup.select("span.text")

    for quote in quotes:
        # print(quote.get_text(strip=True))
        author = quote.find_next("small")
        # print(author.get_text(strip=True))
        data_list.append([quote.get_text(strip=True) , author.get_text(strip=True)])

df = pd.DataFrame(data_list, columns=["quote", "author"])
```

```python
einstein_quotes = df[df["author"] == "Albert Einstein"]
```
