```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
# sns.get_dataset_names()

tips = sns.load_dataset("tips")

# tips.head()
sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip", 	
    hue="smoker",
    size="size",
    # style="smoker"
)
```

```python
flights = sns.load_dataset("flights")

sns.relplot(
    data = flights,
    x = "year",
    y = "passengers", 	
    kind = "line"
)
```

```python
# sns.get_dataset_names()

tips = sns.load_dataset("tips")

# tips.head()
sns.barplot(
    data = tips,
    x = "day",
    y = "tip", 	
    hue = "sex"
)
```

```python
sns.boxplot(
    data = tips,
    x = "day",
    y = "tip", 	
    hue = "sex"
)
```

```python
penguins = sns.load_dataset("penguins")

sns.histplot(
    data = penguins,
    x = "body_mass_g",
    bins = 30
)
```

```python
flights = sns.load_dataset("flights")

flights_pivot = flights.pivot(
    index="month",
    columns="year",
    values="passengers"
)

plt.figure(figsize=(14, 8))

sns.heatmap(
    flights_pivot,
    cmap="YlOrRd",
    annot=True,
    fmt="d",
    linewidths=0.5,
    linecolor="white",
    cbar_kws={
        "label": "Number of Passengers"
    }
)

plt.title(
    "Monthly Air Passengers (1949–1960)",
    fontsize=20,
    fontweight="bold",
    pad=20
)

plt.xlabel("Year", fontsize=13, fontweight="bold")
plt.ylabel("Month", fontsize=13, fontweight="bold")

plt.xticks(rotation=0)
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```
